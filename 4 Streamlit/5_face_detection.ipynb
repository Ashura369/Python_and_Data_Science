# Face Detection Model<br>
Using Open CV<br><br>
**REMEMBER** : while using open cv models, first you have to download them, then you can use it.

```python
import numpy as np
import cv2 as cv
```

### Face detection in video

```python
# for face detection we will be using YuNet model of opencv

vid = cv.VideoCapture(0, cv.CAP_DSHOW)

width = int(vid.get(cv.CAP_PROP_FRAME_WIDTH))
height = int(vid.get(cv.CAP_PROP_FRAME_HEIGHT))

model = cv.FaceDetectorYN.create(
    model='models/face_detection_yunet_2026may.onnx',
    config='',
    input_size=(width, height)
)

name = 'Face Detection Model'
```

```python
while True:
    ret, frame = vid.read()
    if not ret:
        print("ERROR")
        break

    frame = cv.flip(frame, 1)

    # using face detector
    _, faces = model.detect(frame)

    # drawing rectangles around the face
    if faces is not None:
        for face in faces:
            x, y, box_w, box_h = map(int, face[:4])

            cv.rectangle(frame, (x, y), (x + box_w, y + box_h), (0, 255, 0), 2)
            
            # Confidence score (e.g., 0.98)
            score = face[-1]
            cv.putText(frame, f"{score:.2f}", (x, y - 10), cv.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    cv.imshow(name, frame)
    
    if cv.waitKey(1) & 0xFF == ord('x'):
        break

vid.release()
cv.destroyAllWindows()


```

### Face detection in image

```python
import cv2 as cv

# 1. Load image
img = cv.imread("open cv/imgs/face.jpeg")

# 2. Get width and height
h, w = img.shape[:2]

# 3. Create model with image's dimensions (w, h)
model = cv.FaceDetectorYN.create(
    model='models/face_detection_yunet_2026may.onnx',
    config="",
    input_size=(w, h),
    score_threshold=0.6
)

# 4. Save as 'faces' (with an 's')!
_, faces = model.detect(img)

# 5. Loop over 'faces'
if faces is not None:
    for f in faces:
        x, y, box_w, box_h = map(int, f[:4])
        
        # Draw thick green rectangle on the face
        cv.rectangle(img, (x, y), (x + box_w, y + box_h), (0, 255, 0), 8)

        score = f[-1]
        cv.putText(img, f"{score:.2f}", (x, y - 20), cv.FONT_HERSHEY_SIMPLEX, 2.0, (0, 255, 0), 5)

# 6. Show the result
window_name = 'Face Detection Model'
cv.namedWindow(window_name, cv.WINDOW_NORMAL)
cv.resizeWindow(window_name, 800, 600)
cv.imshow(window_name, img)
cv.waitKey(0)
cv.destroyAllWindows()
```
