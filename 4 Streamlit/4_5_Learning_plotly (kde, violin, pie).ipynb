#### `kde`, `violin`, and `pie`

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = px.data.tips()
df.sample(6)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.kdeplot(df, x='total_bill', hue='sex')
plt.show()
```

---

```python
fig = px.violin(df, x='total_bill', y='sex', color='sex')
fig.show()
```

```python
fig = px.violin(df, y='total_bill', height=800, width=700, box=True, points='all')             # plotting the box plot and all the points
fig.show()
```

```python
fig = px.violin(df, y='total_bill', color='smoker', height=600, width=500, violinmode='overlay')             # plotting violinplots one above another
fig.show()  
```

```python
fig = px.pie(df, values='tip', names='day', height=800,
            facet_col='sex', facet_row='time'
)
fig.update_traces(marker=dict(line=dict(color='white', width=3)))
fig.update_layout(template='plotly_dark')
fig.show()
```

```python
fig = px.pie(df, values='tip', names='sex',
            facet_col='time'
)
fig.update_traces(marker=dict(line=dict(color='white', width=3)))
fig.update_layout(template='plotly_dark')
fig.show()
```
