**<h1>Learning Plotly</h1>**
#### `histogram`, `bar`, `line`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
# import dataset
df = pd.read_csv("Final Visualization Dataset.csv")
```

```python
with pd.option_context('display.max_columns',None):
    display(df.sample(5))
```

---

### **Histogram**

```python
fig = px.histogram(df, x='Education Field', color='Education Field', text_auto=True)        # text position shows the count labels
fig.update_traces(textposition='outside')                                                   # shows the count labels outside of histplots
fig.show()
```

```python
fig = px.histogram(df, x='Job Role', color='Job Role', text_auto=True, height=600)
fig.update_traces(textposition='outside')
fig.show()
```

```python
fig = px.histogram(df, x='Education Field', color='Education Field', color_discrete_sequence=['red','blue','black','green','yellow','purple'], text_auto=True)
fig.update_traces(textposition='outside')
fig.show()
```

```python
df['Education Field'].value_counts().reset_index()
```

```python
temp = df['Education Field'].value_counts().reset_index()
px.histogram(temp, x='Education Field', y='count', color='Education Field', text_auto=True)
```

### Using Inbuild Plotly 
### **Barplot**

```python
px.data.gapminder()                         # using Gapminder dataset
```

```python
df = px.data.gapminder()
```

```python
with pd.option_context('display.max_rows',None):
    display(df['country'].value_counts())
```

```python
canada = df[df['country'] == 'Canada']
```

```python
canada
```

```python
fig = px.bar(canada, x='year',y='pop', width=550, height=600, text='year', color='pop',
        hover_data=['lifeExp', 'gdpPercap','continent','country'],                    # the respected feature data will be displayed when featured over the data
        title='Population of Canada Over the Years',
        color_continuous_scale = ['green','yellow','red']                             # setting the color format from green to yellow to red
)

fig.update_layout(title_x=0.5)                                                        # puts the heading of the plot to the center
```

```python
fig = px.bar(canada, x='year',y='lifeExp', width=550, height=600, text='year', color='pop',                 
        labels = {'lifeExp':'Life Expectancy',
                'year':'Year',
                'pop':'Population'
                  },                         # changes the names of the features
        title='Population of Canada Over the Years',
        color_continuous_scale = ['green','yellow','red']                          
)

fig.update_layout(title_x=0.5)    
```

```python
# using groupby -- only contains year and gdpPercap
data = df.groupby('year', as_index=False)['gdpPercap'].mean().round(2)


fig = px.bar(data, x='year', y='gdpPercap', text='gdpPercap',
            title='GDP Per Capital Over the Years',
            width=700,
            color='gdpPercap',
            color_continuous_scale=['green','yellow','red']
)
fig.update_layout(title_x=0.5)
fig.show()
```

### **Lineplot**

```python
px.data.gapminder()
```

```python
df = px.data.gapminder().query("country == 'Canada'")               # .query() -- it can also be used with pandas dataframes, coz it is a pandas command
```

```python
df
```

```python
fig = px.line(df, x='year', y='lifeExp', title='Life Expectancy in Canada',
                labels={
                    'lifeExp' : 'Life Expectancy',
                    'year' : 'Year'
                }
              )
fig.update_layout(title_x=0.5)
fig.show()
```

```python
# checking for countries in each continent

temp = px.data.gapminder()

for i in temp['continent'].unique():
    print(f"{i}  --->  \n{temp[temp['continent'] == i]['country'].unique()}")
    print(f"-------------------------------------------------------------------\n")
```

```python
fig = px.line(temp[temp['continent'] == 'Oceania'], x='year',y='lifeExp', color='country', markers=True)            # if marker is set to true, then little circular data points will be plotted on the line graph
fig.show()
```

```python
fig = px.line(temp[temp['continent'] == 'Americas'], x='year',y='lifeExp', color='country')
fig.show()
```

```python
# USING ANOTHER DATASET
df = px.data.stocks()
df.head()
```

```python
fig = px.line(df, x='date', y='GOOG', markers=True)

# Tell the x-axis to place a tick mark every 12 Months ("M12")
# Tell it to format the text as a 4-digit Year ("%Y")
fig.update_xaxes(dtick="M12", tickformat="%Y")

fig.show()
```

```python
fig = px.line(df, x='date', y='GOOG')

# dtick="M1" places a tick mark every 1 Month
# tickformat="%b %Y" formats the text to look like "Jan 2018"
fig.update_xaxes(dtick="M1", tickformat="%b %Y")

fig.show()
```

```python
# Plotting all the graphs at once
cols = ['GOOG', 'AAPL', 'AMZN', 'FB', 'NFLX', 'MSFT']

fig = px.line(df, x='date', y=cols, title='Tech Stocks Over Time')
fig.update_xaxes(dtick='M1', tickformat="%b %Y")
fig.show()
```

```python
# Plotting all the graphs one after another
cols = ['GOOG', 'AAPL', 'AMZN', 'FB', 'NFLX', 'MSFT']

for i in cols:
    fig = px.line(df, x='date', y=i, title=i)
    fig.update_xaxes(dtick='M1', tickformat="%b %Y")
    fig.show()
```
