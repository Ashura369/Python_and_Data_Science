# Face Detection Model (YOLO26)<br>
Using `YOLO26` by `ultralytics`

```python
import numpy as np
import cv2 as cv
from ultralytics import YOLO
```

```python
# pre-trained model (yolo26)
model = YOLO("yolo26n.pt")
```

#### `yolo26n` on live webcam

```python
cap = cv.VideoCapture(0, cv.CAP_DSHOW)
width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))


while True:
    ret, frame = cap.read()

    if not ret:
        print("ERROR")
        break
    

    results = model(frame)

    annotated_frame = results[0].plot()

    cv.imshow("Object Detection Model", annotated_frame)

    if cv.waitKey(1) & 0xFF == ord('x'):
        break

cap.release()
cv.destroyAllWindows()
```

#### `yolo26n` on imgs

```python
img = cv.imread("open cv/imgs/img (7).jpg")

name = 'image'

cv.namedWindow(name, cv.WINDOW_NORMAL)
cv.resizeWindow(name, 800,600)

result = model(img)

cv.imshow(name, result[0].plot())
cv.waitKey(0)
cv.destroyAllWindows()
```
