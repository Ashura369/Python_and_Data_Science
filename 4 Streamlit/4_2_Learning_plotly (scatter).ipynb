### `scatter`

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = px.data.iris()
df.sample(6)
```

```python
df['species_id'].unique()
```

```python
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species_id')
# it will plot species_id in continuous colors

fig.show()
```

```python
fig = px.scatter(df, x='sepal_width', y='sepal_length', color=df['species_id'].astype(str))
# it will plot species_id in fixedcolors

fig.show()
```

```python
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species', size='petal_length')

fig.show()
```

```python
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species', size='petal_length', symbol='species')

fig.show()
```

```python
df = px.data.tips()
df
```

```python
# using 'facet'
fig = px.scatter(df, x='total_bill', y='tip', 
                color='smoker', facet_col='sex', title='Facet Col'
)
fig.show()
```

```python
fig = px.scatter(df, x='total_bill', y='tip', 
                color='smoker', facet_row='time', title='Facet Row', height=600
)
fig.show()
```

```python
fig = px.scatter(df, x='total_bill', y='tip', 
                color='smoker', facet_col='sex', facet_row='day', title='Facet Col and Facet Row', 
                height=600
)
fig.show()
```

```python
df
```

```python
# using Marginal Plot
    # REMEMBER BOTH marginal_x and marginal_y can take only 4 values : histogram, rug, box, violin

fig = px.scatter(df, x='tip', y='total_bill', color='smoker',
                marginal_x='histogram', marginal_y='rug',
                title='Scatter Plot with Histogram and Rug',
                height=700
)

fig.show()
```

```python
# using Marginal Plot
fig = px.scatter(df, x='tip', y='total_bill', color='smoker',
                marginal_x='box', marginal_y='violin', trendline='ols',
                title='Scatter Plot with Boxplot and Violinplot',
                height=700
)

fig.show()
```

**REMEMBER** : `marginal_x` and `marginal_y` are only limited to scatterplot, density_headmap, density_contur.<br>
While using histogram, it can only take one marginal.
