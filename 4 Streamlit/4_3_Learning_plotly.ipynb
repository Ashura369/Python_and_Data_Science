### `histogram`

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = px.data.tips()
df.sample(5)
```

```python
fig = px.histogram(df, x='total_bill')
fig.show()
```

```python
fig = px.histogram(df, x='total_bill', nbins=20)            # putting bins
fig.update_layout(bargap=0.2)                               # puts gap btwn bars
fig.show()
```

```python
fig = px.histogram(df, x='day')
fig.show()
```

```python
# using Histnorm

fig = px.histogram(df, x='total_bill', histnorm=None)          # histnorm changes what y axis of histogram represents
# histnorm takes only 5 values
    # None -- by default takes 'count'
    # percent
    # probability
    # density
    # probability density

fig.show()
```

```python
# using Histfunc

fig = px.histogram(df, x='total_bill', y='tip', histfunc='avg')         # make the y axis into avg
# the values histfunc takes
    # count
    # avg
    # sum
    # min
    # max
    
fig.show()

```

```python
stocks = px.data.stocks()
stocks.sample(5)
```

```python
fig = px.histogram(stocks, x='date')
fig.update_layout(bargap=0.2)
fig.show()
```
