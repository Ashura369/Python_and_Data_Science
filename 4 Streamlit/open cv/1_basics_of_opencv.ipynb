# Basics of `Open CV`

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("imgs/img (2).jpg")                         # returns a numpy array for the image

print(f"Type of Img : {type(img)}")
print(f"Shape of array : {img.shape}")
print(f"Array format for img : {img}")
```

### Making a pop-up screen for the img

```python
cv2.imshow("Image 1", img)

cv2.waitKey(0)
# The 0 means "wait forever" until you press any key on your keyboard (like Space, Enter, or Esc). As soon as you hit a key, the program resumes and closes the window.
# The number is in milliseconds. 3000 means show the image for 3 seconds, then close automatically without waiting for you.

cv2.destroyAllWindows()
# completely closes and removes all popup image/video windows created by cv2.imshow() and frees up memory from your operating system
```

### Making a pop-up window with specific width and height size

```python
cv2.namedWindow('Img 1', cv2.WINDOW_NORMAL)             # creates resizable window with the same name as 'imshow'
cv2.resizeWindow('Img 1', 800, 500)                     # setting the custorm window size 800x600
cv2.imshow('Img 1', img)
cv2.waitKey(2000)
cv2.destroyAllWindows()
```

### Printing the img using `Matplotlib`

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

### Converting an img into grayscale img

```python
img = cv2.imread("imgs/img (2).jpg")

# 1. Name must match across namedWindow, resizeWindow, and imshow
window_name = 'Grayscale img'

cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
cv2.resizeWindow(window_name, 800, 500)         # 800 height, 500 width

img_grayscale = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow(window_name, img_grayscale)

# 2. Fixed capital 'K' in waitKey
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Plotting 2 images side by side

```python
img = cv2.imread("imgs/img (3).jpg")

name = 'Img 3'
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_gray_3ch = cv2.cvtColor(img_gray, cv2.COLOR_GRAY2BGR)               # converting 1-channel grayscale to 3-channel so dimensions match!


combined = np.hstack((img, img_gray_3ch))

cv2.namedWindow(name, cv2.WINDOW_NORMAL)
cv2.resizeWindow(name, 1400,500)

cv2.imshow(name, combined)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Flipiping imgs

```python
img = cv2.imread("imgs/img (7).jpg")

name = 'img 7'
cv2.namedWindow(name, cv2.WINDOW_NORMAL)
cv2.resizeWindow(name, 1200, 800)

# there are three types of flip values 
# 0 -- Flips the image top-to-bottom across the horizontal (X) axis
# 1 -- Flips the image left-to-right across the vertical (Y) axis
# -1 -- Flips both vertically and horizontally at the same time
img_flip = cv2.flip(img, 1)
cv2.imshow(name, img_flip)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Cropping imgs

```python
img = cv2.imread("imgs/img (7).jpg")
names = 'img 7'

cv2.imshow(names, img[:1000, 0:800])
cv2.waitKey()
cv2.destroyAllWindows()
```

### Saving imgs

```python
img = cv2.imread("imgs/img (7).jpg")
name = 'img 7'

cropped_img = img[:1000, 0:800]

# saving the img
cv2.imwrite("imgs/Saved Img.jpg", cropped_img)

cv2.waitKey()
cv2.destroyAllWindows()
```
