# Drawing Shapes (Rectangles, Circles, Lines) and texts <br>
**IMP FOR OBJECT DETECTION**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Rectangle

```python
# making a black img
# using np.zeros coz 0 is for black and 1 is for white
img = np.zeros((512,512,3))
cv2.rectangle(img, (100,100), (300,300), (255,0,0), 2)                  # # Top-Left: (100, 100), Bottom-Right: (300, 300), Color: (255, 0, 0), Thickness: 3
cv2.rectangle(img, (200,200), (400,400), (0,255,0), 2)                 
cv2.rectangle(img, (300,300), (500,500), (0,0,255), 2)     
cv2.rectangle(img, (50,50), (100,100), (0,0,255), -1)                   # if -1 is written, it will fil the rectangle box     

# blue = (255,0,0)
# green = (0,255,0)
# red = (0,0,255)

cv2.namedWindow('img', cv2.WINDOW_NORMAL)
cv2.resizeWindow('img', 500, 500)                          # 500 - width, 1000 - height

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Circle

```python
# making a black img
img = np.zeros((512,512,3))
cv2.circle(img, (100,300), 50, (255,0,0), 2)                  # Center=(100, 300), Radius=50, Color=(255, 0, 0), Thickness=2
                                                                # 100 is horizontal, 300 is vertical
cv2.circle(img, (300,100), 50, (0,255,0), -1) 
cv2.rectangle(img, (200,200),(400,400), (0,0,255), 1)


cv2.namedWindow('img', cv2.WINDOW_NORMAL)
cv2.resizeWindow('img', 500, 500)                         

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Line

```python
img = np.zeros((512,512,3))
cv2.line(img, (200,200),(400,400), (0,0,255), 2)                   

cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Text

```python
import cv2
import numpy as np

img = np.zeros((512, 512, 3), dtype=np.uint8)

# 2. Draw text (Clean positional format)
# cv2.putText(img, text, org, fontFace, fontScale, color, thickness, lineType)
cv2.putText(img, 'Hello', (50, 250), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 0, 255), 2, cv2.LINE_AA)

# 3. Show the image window!
cv2.imshow('My Window', img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
