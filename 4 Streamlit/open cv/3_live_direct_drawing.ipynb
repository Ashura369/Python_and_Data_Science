# Live drawing

### Making circles appear on the screen based on mouse clicks

```python
import numpy as np
import cv2
```

```python
img = np.ones((512,512,3))
cv2.imshow('img', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Drawing squares and circles

```python
red = (0,0,255)
green = (0,255,0)
blue = (255,0,0)



def draw(event, x, y, flags, params):        
    if event == cv2.EVENT_LBUTTONDOWN:
        print("Left Click!")
        cv2.rectangle(img, (x,y), (x+50, y+50), red, 2)
    elif event == cv2.EVENT_RBUTTONDOWN:
        print("Right Click!")
        cv2.circle(img, (x,y), 25, green, -1)
    elif event == cv2.EVENT_LBUTTONDBLCLK:
        print("Double Clicked!")
        img[:] = 1                  # resetting img back to normal


cv2.namedWindow(winname='window')
cv2.setMouseCallback('window', draw)        


# making a loop of img
while True:
    cv2.imshow('window', img)
    if cv2.waitKey(1) & 0xFF == ord('x'):       # the img will kepp appearing until you press 'x'
        break

cv2.destroyAllWindows()


```

### 📋 OpenCV Mouse Events Cheat Sheet

| Event Code | OpenCV Constant Name | User Mouse Action |
| :---: | :--- | :--- |
| **0** | `cv2.EVENT_MOUSEMOVE` | Moving the mouse cursor over the window |
| **1** | `cv2.EVENT_LBUTTONDOWN` | Pressing down the Left mouse button |
| **2** | `cv2.EVENT_RBUTTONDOWN` | Pressing down the Right mouse button |
| **3** | `cv2.EVENT_MBUTTONDOWN` | Pressing down the Middle (Scroll Wheel) button |
| **4** | `cv2.EVENT_LBUTTONUP` | Releasing / letting go of the Left mouse button |
| **5** | `cv2.EVENT_RBUTTONUP` | Releasing / letting go of the Right mouse button |
| **6** | `cv2.EVENT_MBUTTONUP` | Releasing / letting go of the Middle button |
| **7** | `cv2.EVENT_LBUTTONDBLCLK` | Double-clicking the Left mouse button |
| **8** | `cv2.EVENT_RBUTTONDBLCLK` | Double-clicking the Right mouse button |
| **9** | `cv2.EVENT_MBUTTONDBLCLK` | Double-clicking the Middle mouse button |
| **10** | `cv2.EVENT_MOUSEWHEEL` | Scrolling the mouse wheel vertically (Up/Down) |
| **11** | `cv2.EVENT_MOUSEHWHEEL` | Scrolling the mouse wheel horizontally (Left/Right) |

### Making random shapes

```python
import cv2
import numpy as np


#  Tracking variables
flag = False
ix = -1
iy = -1

#  Mouse Callback function
def draw(event, x, y, flags, params):   
    global flag, ix, iy

    # When left click is pressed down -> Start dragging
    if event == cv2.EVENT_LBUTTONDOWN:
        flag = True
        ix = x
        iy = y

    # When mouse moves -> If dragging, draw rectangle
    elif event == cv2.EVENT_MOUSEMOVE:
        if flag == True:
            cv2.rectangle(img, pt1=(ix, iy), pt2=(x, y), color=red, thickness=-1)

    # When left click is released -> Stop dragging
    elif event == cv2.EVENT_LBUTTONUP:
        flag = False
        cv2.rectangle(img, pt1=(ix, iy), pt2=(x, y), color=red, thickness=-1)

    elif event == cv2.EVENT_LBUTTONDBLCLK:
        img[:] = 1

#  Window setup
cv2.namedWindow(winname='window')
cv2.setMouseCallback('window', draw)        

#  Main loop
while True:
    cv2.imshow('window', img)
    # Press 'x' on keyboard to exit
    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

cv2.destroyAllWindows()
```
