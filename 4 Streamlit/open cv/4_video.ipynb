# Video in Open CV

```python
import numpy as np
import cv2
```

```python
vid = cv2.VideoCapture(0, cv2.CAP_DSHOW)           # if you a file name instead of 0, you will be able to run a video file, and delete the "cv2.CAP_DSHOW"
name = 'Live Webcam Feed'

while True:
    ret, frame = vid.read()
    if not ret:
        print("Camera not returning frames")
        break

    # cv2.namedWindow(name, cv2.WINDOW_NORMAL)
    # cv2.resizeWindow(name, 1000,800)

    # frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)             # for grayscale video

    frame = cv2.flip(frame, 1)              # for mirror video feed
    cv2.imshow(name, frame)
    
    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

vid.release()               # disconnects from the camera hardware and closes the video file
cv2.destroyAllWindows()
```

### `ret, frame = vid.read()` in OpenCV
---

When you call `vid.read()`, OpenCV captures a single snapshot from the video stream and returns a tuple of **two values**:


#### 1. `ret` (Return / Status Flag)
* **Data Type:** `bool` (`True` or `False`)
* **`True`**: Frame was successfully captured and decoded.
* **`False`**: No frame could be retrieved.
  * **Video File:** Reached the end of the video.
  * **Webcam:** Camera disconnected, busy, or permission denied.

> **Why we use `if not ret: break`:** Prevents Python from crashing when attempting to process a `None` frame at the end of the stream.


#### 2. `frame` (The Image Matrix)
* **Data Type:** `numpy.ndarray` (`dtype=uint8`)
* **Color Space:** `BGR` (Blue, Green, Red) with pixel intensity `[0, 255]`.
* **Shape:** `(Height, Width, Channels)` — e.g., `(480, 640, 3)`.
* **Note:** If `ret == False`, `frame` will be `None`.



#### Summary 

| Return Variable | Type | Value / Content | Description |
| :--- | :---: | :--- | :--- |
| **`ret`** | `bool` | `True` or `False` | Status flag confirming frame capture success |
| **`frame`** | `ndarray` | `(H, W, 3)` Array | The actual image/pixel matrix of the captured frame |


### Saving the web cam video feed

```python
vid = cv2.VideoCapture(0, cv2.CAP_DSHOW)  

# getting the camera's natural width and height
width = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))


# Defining the Codec and create the VideoWriter object
# 'mp4v' is the standard codec for .mp4 files
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

# Arguments: (filename, fourcc_codec, fps, (width, height))
out = cv2.VideoWriter('Recorded video..mp4', fourcc, 30.0, (width, height))


name = 'Live Video Feed'

while True:
    ret, frame = vid.read()
    if not ret:
        print("Camera not returning frames")
        break

    # cv2.namedWindow(name, cv2.WINDOW_NORMAL)
    # cv2.resizeWindow(name, 1000,800)

    # frame = cv2.flip(frame, 1)

    out.write(frame)                # writing the current frame to the disk
    cv2.imshow(name, frame)
    
    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

vid.release()         
out.release()      
cv2.destroyAllWindows()
```
