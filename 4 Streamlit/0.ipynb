# `Streamlit` Notes
---

`st.text()`
- Purpose: Used for displaying plain text exactly as written.
- Formatting: Renders text in a monospace font (fixed-width, like a code block or terminal output).
- Flexibility: Only accepts strings. It cannot parse Markdown, emojis, dataframes, charts, or other objects.
- Performance: Slightly faster because it is lightweight and doesn't need to inspect or evaluate the data type.

`st.write()`
- Purpose: Streamlit's "Swiss Army knife" function. It can render text, Markdown, dataframes, charts, Matplotlib figures, and more.
- Formatting: Renders standard text using the app's default proportional font. Supports full Markdown syntax (bold, italics, headers, lists).
- Flexibility: Automatically detects the input data type (e.g., if you pass a Pandas DataFrame, it renders an interactive table; if you pass a dictionary, it renders a JSON viewer).
- Performance: Slightly heavier because it evaluates the data type before rendering.
