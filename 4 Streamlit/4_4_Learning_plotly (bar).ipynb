#### `box`

```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = px.data.tips()
df.sample(6)
```

```python
fig = px.box(df, x='tip', color='sex')
fig.show()
```

```python
fig = px.box(df, x='tip', color='smoker')
fig.show()
```

```python
fig = px.box(df, y='total_bill', x='sex', color='sex', height=500, width=800, points='all')         # points='all' -- plots the distributions of all the data points
fig.show()
```

```python
fig = px.box(df, x='total_bill', y='sex', color='sex', notched=True)
fig.show()
```
