In Plotly, understanding the difference between the **core figure creation** (like `px.histogram` or `px.scatter`) and methods like `.update_layout()` or `.update_traces()` is the secret to moving from basic charts to fully customized, professional dashboards.

Here is a breakdown of what those methods are, followed by the other major structural tools you will use in Plotly.


#### 1. What are `fig.update_layout()` and `fig.update_traces()`?

Think of a Plotly figure as a nested JSON structure made of two main parts: the **Canvas/Container** and the **Data Objects (Traces)**.

* **`fig.update_layout()` (The Canvas & Global Settings):**
  * **What it controls:** Everything *around* the data. This includes the background, overall title, legend positioning, axes titles, ranges, font styles, and margins. 
  * **Analogy:** It’s like buying a picture frame, setting the wall color, and adjusting the room lighting.
  * **Example:** `fig.update_layout(title="My Chart", template="plotly_dark", xaxis_title="Years")`

* **`fig.update_traces()` (The Data Objects inside the canvas):**
  * **What it controls:** The actual visual representations of your data points, bars, lines, or markers. This is where you style how the data itself looks (like adding the bar border lines we just discussed).
  * **Analogy:** It’s like painting the individual objects inside the picture frame (e.g., making all the cars red).
  * **Example:** `fig.update_traces(marker_color='orange', opacity=0.8)`


#### 2. Other Powerful `fig.update_...` Methods You Should Know

Plotly provides a whole family of customization methods that follow this exact pattern. Here are the other essential ones you will use:

#### A. `fig.update_xaxes()` & `fig.update_yaxes()`
Instead of digging through layout settings, these give you direct control over your axes (gridlines, tick marks, log scales, or reversing axes).
* **Example:** 
  ```python
  fig.update_xaxes(showgrid=False, tickangle=45) # Hides vertical grid lines and tilts labels

---

- `fig.add_annotation()`<br>
Allows you to add custom text notes, pointers, or highlight specific data points directly on the chart canvas.
```python 
    fig.add_annotation(x=5, y=10000, text="Promotion Bottleneck", showarrow=True)
```

- `fig.add_shape() / fig.add_hline() / fig.add_vline()`<br>
Perfect for adding threshold lines (like a company average line or a risk cutoff threshold) onto your plots.
```py
    fig.add_hline(y=5000, line_dash="dash", line_color="red", annotation_text="Target Income")

```

- `fig.update_legends()`<br>
Controls how legends behave—whether they sit at the top, bottom, inside the chart, or get hidden entirely.
```py
    fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1))
```



````python
df = px.data.tips()
df.sample(5)
````

````python
fig = px.histogram(df, x='total_bill', y='tip', nbins=30, color='day',
                    facet_row='day', height=800
                   )

fig.update_traces(
    marker=dict(
        line=dict(color='black', width=1)
    )
)

fig.update_layout(title="Total Bill Chart", title_x=0.5, template="plotly_dark", xaxis_title='Total Bill', yaxis_title='Count of Total Bill', )



fig.show()
````

````python
fig = px.histogram(df, x='total_bill', nbins=20, width=1000, 
                   facet_col='smoker', 
                   labels={
                       'total_bill' : 'Total Bill'
                   }

                   )
fig.update_traces(marker=dict(
    line=dict(color='black', width=2)
))
fig.update_layout(bargap=0.2, template='plotly_white')                               # puts gap btwn bars
fig.show()
````

````python
fig = px.histogram(df, x='day')

fig.update_layout(template='plotly_dark')

fig.show()
````

````python
# using Histnorm

fig = px.histogram(df, x='total_bill', width=800, histnorm=None)          # histnorm changes what y axis of histogram represents
# histnorm takes only 5 values
    # None -- by default takes 'count'
    # percent
    # probability
    # density
    # probability density
fig.update_traces(marker=dict(line=dict(color='black', width=2)))
fig.update_layout(bargap=0.2, template='plotly_dark')
fig.show()
````

````python
# using Histfunc

fig = px.histogram(df, x='total_bill', y='tip', histfunc='avg')         # make the y axis into avg
# the values histfunc takes
    # count
    # avg
    # sum
    # min
    # max

fig.update_traces(marker=dict(line=dict(color='black', width=2)))
fig.update_layout(bargap=0.2, template='plotly_dark')
fig.show()

````

````python
# using Histfunc

fig = px.histogram(df, x='total_bill', y='tip', histnorm='probability density', text_auto=True)       

fig.update_traces(
        textposition='outside',
        marker=dict(line=dict(color='black', width=2))
    )
fig.update_layout(template='plotly_dark', bargap=0.2)
fig.show()

````
